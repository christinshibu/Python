```python
#Return Statement 10 Practice Questions
#1.Write a function add() that returns the sum of two numbers.

def add(a,b):
    result=a+b
    return result

value=add(5,3)
print(value)

#2.Write a function square() that takes a number and returns its square.

def square(num):
    result=num*num
    return result

value=square(4)
print(value)

#3.Write a function multiply() that takes two numbers and returns their product.

def multiply(a,b):
    result=a*b
    return result

value=multiply(6,7)
print(value)

#4.Write a function is even() that takes a number and returns True if it is even, otherwise False.

def even(num):
    result=(num%2==0)
    return result

value=even(10)
print(value)

#5.Write a function find max() that takes two numbers and returns the greater euaber.

def find_max(a,b):
    result=a if a>b else b
    return result

value=find_max(15, 20)
print(value)

#6.Write a function get full name() that takes first name and last name and returns the complete name.

def full_name(first,last):
    result = first+" "+last
    return result

value=full_name("Ram","S")
print(value)

#7.Write a function calculate_area() that takes the length and breadth of a rectangle and returns its area.

def calculate_area(length, breadth):
    result=length*breadth
    return result

value=calculate_area(10, 5)
print(value)

#8.Write a function calculate_average() that takes three mumbers and returns their average.

def calculate_average(a,b,c):
    result = (a+b+c)/3
    return result

value=calculate_average(10, 20, 30)
print(value)

#9.Write a function get grade() that takes marks and returns:
#"A" if marks 2 90
#"8" if marks 275
#"C" if marks 2 50

def grade(marks):
    if marks>=90:
        result="A"
    elif marks>=75:
        result="B"
    elif marks>=50:
        result="C"
    else:
        result="Fail"
    return result

value=grade(92)
print(value)

value=grade(76)
print(value)

value=grade(55)
print(value)

value=grade(40)
print(value)

#10.Write a function calculate total() that takes the prices of three itens and returns the total price. 
# Then store the returned value in a variable and print it

def calculate_total(p1,p2,p3):
    result=p1+p2+p3
    return result

total_price=calculate_total(100,200,300)
print(total_price)
```
