```python
# 6.Rectangle Calculation
# Create a class Rectangle with a constructor that accepts length and breadth. 
# Create methods to calculate and display the area and perimeter

class Rectangle:
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth
    def calculate(self):
        area=self.length*self.breadth
        perimeter=2*(self.length+self.breadth)
        print("Area of the rectangle is",area)
        print("Perimeter of the rectangle is",perimeter)
        
result=Rectangle(3,4)
result.calculate()
```

```python
# Q7. Circle Calculation
# Create a class Circle with a constructor that accepts radius. 
# Create methods to calculate the area and circumference.

class Circle:
    def __init__(self,radius):
        self.radius=radius
    def calculate(self):
        area=3.14*self.radius*self.radius
        circumference=round((2*3.14*self.radius),1)
        print("Area of the circle is",area)
        print("Circumference of the circle is",circumference)
result=Circle(5)
result.calculate()
```

```python
# Q9. Student Marks
# Create a class Student with a constructor that accepts name, mark1, mark2 and mark3. 
# Create methods to calculate the total and average marks.

class Student:
    def __init__(self,name,mark1=20,mark2=40,mark3=35):
        self.name=name
        self.mark1=mark1
        self.mark2=mark2
        self.mark3=mark3
    def display(self):
        print("Name of the student is :",self.name)
        print("Marks obtained for subject 1:",self.mark1)
        print("Marks obtained for subject 2:",self.mark2)
        print("Marks obtained for subject 3:",self.mark3)
    def calculate(self):
        total=self.mark1+self.mark2+self.mark3
        average=round((self.mark1+self.mark2+self.mark3)/3,2)
        print("Total marks:",total)
        print("Average marks:",average)
result=Student("Sam")
result.display()
result.calculate()
```

```python
# Q10. Product Discount
# Create a class Product with a constructor that accepts name and price. Create a 
# method to calculate a 10% discount and another method to display the final price. 

class Product:
    def __init__(self,name,price=100):
        self.name=name
        self.price=price
    def calculate(self):
        discount=(round(self.price*0.10,2))
        final_price=self.price-discount
        print("Name of the book :",self.name)
        print("Price of the book :",self.price)
        print("Discounted price :",final_price)
result=Product("Alchemist")
result.calculate()
```

```python
# Q18. Constructor and Multiple Operations
# Create a class BankAccount with a constructor that accepts name and balance. 
# Create methods to:
# Deposit money.
# Withdraw money, provided the balance is sufficient.
# Display the current balance.

class BankAccount:
  def __init__(self,name,balance):
    self.name = name
    self.balance = balance

  def deposit(self,amount):
    self.balance +=amount

  def withdraw(self,amount):
    if(amount>self.balance):
      print("Insufficient balance")
    else:
      self.balance -= amount

  def display(self):
    print(self.name)
    print(self.balance)

b = BankAccount("Adharv",1000)
b.deposit(10000)
b.withdraw(2000)
b.display()
```

```python
# Q19. Compare Two Objects
# Create a class Product with a constructor that accepts name and price. 
# Create two objects and a method that compares their prices and displays which 
# product is more expensive.

class Product:
  def __init__(self,name,price):
    self.name = name
    self.price = price
  def compare(self, other):
    if(other.price>self.price):
      print(f"{other.name} is more expensive")
    else:
      print(f"{self.name} is more expensive")

p = Product("Car",100000)
d = Product("Bike",50000)
p.compare(d)
```

```python
# Q20. Library Management
# Create a class LibraryBook with a constructor that accepts book_name, 
# author, and availability. Create methods to:
# Display book details.
# Issue a book if available.
# Return a book.
# Display the current availability status

class LibraryBook:
  def __init__(self, book_name, author, availability):
    self.book_name = book_name
    self.author = author
    self.availability = availability
  def display(self):
    print(self.book_name)
    print(self.author)
    print(self.availability)
  def issue(self):
    if(self.availability):
      self.availability = False
      print("Book Issued")
    else:
      print("Book not available")

  def return_book(self):
    if(not self.availability):
      self.availability = True
      print("Book returned")
    else:
      print("Book not returned")

result=LibraryBook("Alchemist","Paulo Ceohlo",1)
result.display()
result.issue()
result.return_book()
```
