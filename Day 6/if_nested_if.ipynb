```python
#3. Wipro - For 225, check whether it is positive. if positive, check whether it is odd. 
# If odd, check whether it is divisible by 3,5,or 15 using nested if-elif-else

a=225
if (a>0):
    print(a,"is positive")
    if(a%2==1):
        print("it is odd")
        if(a%3==0):
            print(a,"is divisible by 3")
        elif(a%5==0):
            print(a,"is divisible by 5")
        elif(a%15==0):
            print(a,"is divisible by 15")
else:
    print(a,"is negative")
```
