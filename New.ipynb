```python
a=[1,2,[3,8],4,5,6]

b=a.copy() #shallow
a[2][1]=1000
print(a)
print(b)





```

```python
import copy
a=[10,20,[30,40]]
b=copy.deepcopy(a)

b[2][0]=90

print("Original",a)
print("Deep copy",b)

```
