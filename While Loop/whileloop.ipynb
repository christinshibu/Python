```python
# 47. Check whether a number is an Armstrong number.

n = int(input("Enter a number: "))

temp = n
sum = 0

while temp > 0:
    digit = temp % 10
    sum = sum + digit ** 3
    temp = temp // 10

if sum == n:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```

```python
# 48. Print Armstrong numbers from 1 to n.

n = int(input("Enter the limit: "))

num = 1

while num <= n:
    temp = num
    sum = 0

    while temp > 0:
        digit = temp % 10
        sum = sum + digit ** 3
        temp = temp // 10

    if sum == num:
        print(num)

    num += 1
```

```python
# 49. Check whether a number is a strong number.

n = int(input("Enter a number: "))
temp = n
sum = 0
while temp > 0:
    digit = temp % 10
    i = 1
    fact = 1
    while i <= digit:
        fact = fact * i
        i += 1
    sum = sum + fact
    temp = temp // 10
if sum == n:
    print("Strong number")
else:
    print("Not a strong number")
```

```python
# 50. Print strong numbers within a range.

start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))
num = start
while num <= end:
    temp = num
    sum = 0
    while temp > 0:
        digit = temp % 10
        i = 1
        fact = 1
        while i <= digit:
            fact = fact * i
            i += 1
        sum = sum + fact
        temp = temp // 10
    if sum == num:
        print(num)
    num += 1
```
