```python
class Parent1():
    def __init__(self,p1_name):
        self.p1_name=p1_name
        print(p1_name)

class Parent2():
    def __init__(self,p2_name):
        self.p2_name=p2_name
        print(p2_name)
class Child(Parent1,Parent2):
    def __init__(self,c_name):
        Parent1.__init__(self,"P1")
        Parent2.__init__(self,"p2")
        self.c_name=c_name
        print(c_name)
p1=Parent1("Parent 1")
p2=Parent2("Parent 2")
c=Child("Child")
```

```python
class Parent1():
    def __init__(self,p1_name,p1_work,p1_age):
        self.p1_name=p1_name
        self.p1_work=p1_work
        self.p1_age=p1_age
        print("Parent 1")
        print("Parent 1 Name",p1_name)
        print("Parent 1 work",p1_work)
        print("Parent 1 age",p1_age)
class Parent2():
    def __init__(self,p2_name,p2_work,p2_age):
        self.p2_name=p2_name
        self.p2_work=p2_work
        self.p2_age=p2_age
        print("Parent 2")
        print("Parent 2 Name",p2_name)
        print("Parent 2 work",p2_work)
        print("Parent 3 age",p2_age)
class Sibling():
    def __init__(self,s_name,s_age,s_grade):
        self.s_name=s_name
        self.s_age=s_age
        self.s_grade=s_grade
        print("Sibling")
        print("Sibling Name",s_name)
        print("Sibling age",s_age)
        print("Sibling grade",s_grade)
class MyDetails(Parent1,Parent2,Sibling):
    def __init__(self,name,age,qualification):
        Parent1.__init__(self,"abc","ggg",90)
        Parent2.__init__(self,"gg","hh",80)
        Sibling.__init__(self,"yy",45,"A")
        self.name=name
        self.age=age
        self.qualification=qualification
        
        print("My Details")
        print("My Name",name)
        print("My age",age)
        print("My Qualification",qualification)
p1=Parent1("abc","ggg",90)
p2=Parent2("gg","hh",80)
s1=Sibling("yy",45,"A")
m=MyDetails("Cc",23,"MCA")
```

```python
# 10. create a user-defined exception for checking student marks.

class InvalidMarksError(Exception):
    pass

try:
    marks = int(input("Enter student marks: "))

    if marks < 0 or marks > 100:
        raise InvalidMarksError("Invalid marks! Enter marks between 0 and 100")

    print("Student marks:", marks)

except InvalidMarksError as e:
    print(e)

except ValueError:
    print("Please enter a valid number")
```
