```python
#3. Take one integer and determine whether it is a single-digit, two-digit, 
# three-digit, or more-than-three-digit number.

a=100
if(a>-10 and a<10):
    print(a,"is single digit")
elif(a>9 and a<100):
    print(a,"is two digit")
elif(a>99 and a<1000):
    print(a,"is three digit")
else:
    print(a,"is above three digit")
```
